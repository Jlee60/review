## Chapter 2 – Types and Variables

This chapter explains how computers store data using **bits**, **bytes**, and larger multi‑byte values, and how
Python represents information with different **types**. You also learn how to give data a name with a **variable**
so that you can reuse it later in your program.

**Key ideas from Chapter 2:**  
- Bits and bytes as low‑level building blocks of data  
- Python values have types (integers, floats, strings, and more)  
- A **variable** is a name that refers to a value in memory  
- You can assign, change, and delete variables with `=`, reassignment, and `del`  
- Rules and conventions for naming variables


```python
# Chapter 2 – types and variables

# Python figures out the type from the value on the right‑hand side
age = 20          # an integer
temperature = 21.5  # a float
message = "Hello"   # a string

print("age:", age)
print("temperature:", temperature)
print("message:", message)

# Reassignment: the name can later refer to a different value
age = age + 1  # add 1 to the previous value
print("age after a birthday:", age)

# Variables are labels for values in memory
a = 10
b = a          # b refers to the same integer value as a
print("a:", a, "b:", b)

# Changing a just makes the name a point to a new value
a = 99
print("after changing a -> a:", a, "b is still:", b)

# You can delete a variable name with del
pet = "dog"
print("pet before deleting:", pet)
del pet
# After deletion, using the name pet would cause an error.

# Try making your own variables following Python naming rules,
# such as: letters, numbers, and underscores, but not starting with a number.
```

## Chapter 3 – Numbers

This chapter focuses on Python’s basic numeric types and how to do math with them.
You explore **booleans**, **integers**, and **floating‑point numbers**, learn about operator precedence,
and see how to convert between number types.

**Key ideas from Chapter 3:**  
- Boolean values `True` and `False` and basic comparisons  
- Integers for whole numbers and floats for numbers with a decimal point  
- Common arithmetic operators: `+`, `-`, `*`, `/`, `//`, `%`, `**`  
- Precedence (order) of operations and the use of parentheses  
- Converting between types with `int()` and `float()`  
- Using the `math` module for extra numeric functions


```python
# Chapter 3 – working with numbers

# Booleans come from comparisons
print("3 > 1 ?", 3 > 1)
print("2 == 5 ?", 2 == 5)

# Integers and floats
apples = 5            # integer
price_per_apple = 0.75  # float

total_cost = apples * price_per_apple
print("Total cost:", total_cost)

# Integer division (//) and remainder (%)
students = 23
group_size = 4
full_groups = students // group_size
left_over = students % group_size

print("Full groups:", full_groups)
print("Students left over:", left_over)

# Precedence: * happens before +
result1 = 2 + 3 * 4      # 2 + (3*4)
result2 = (2 + 3) * 4    # (2+3) * 4
print("result1:", result1, "result2:", result2)

# Type conversions
pi_approx = 3.14
print("float to int:", int(pi_approx))   # drops the fraction
print("int to float:", float(10))

# Using the math module for extra functions
import math

print("Square root of 16:", math.sqrt(16))
print("Cosine of 0:", math.cos(0))

# Try changing the numbers or operations and run again.
```

## Chapter 4 – Strings

This chapter explores **text strings**, which are sequences of characters. You learn how to create them,
combine them, slice out pieces, and format them for display.

**Key ideas from Chapter 4:**  
- Creating strings with quotes or with `str()`  
- Concatenation with `+` and repetition with `*`  
- Indexing and slicing to get characters and substrings  
- String functions like `len()`, `split()`, `join()`, and `replace()`  
- Checking prefixes/suffixes and changing case  
- Formatting values inside strings, including f‑strings


```python
# Chapter 4 – using strings

# Creating strings
greeting = "Hello"
name = "Ada"
print(greeting, name)

# Concatenation and repetition
full_message = greeting + ", " + name + "!"
print(full_message)
print("ha" * 3)  # repeat the string

# Indexing and slicing
word = "Python"
print("First letter:", word[0])
print("Last letter:", word[-1])
print("First three letters:", word[:3])
print("Middle slice:", word[1:5])

# Length of a string
print("Length of", word, "is", len(word))

# Splitting and joining
sentence = "red,green,blue"
colors = sentence.split(",")   # split into a list
print("colors list:", colors)

joined = " & ".join(colors)    # join back into a string
print("joined:", joined)

# Replacing text
print("replace 'red' with 'yellow':", sentence.replace("red", "yellow"))

# Case methods and f‑strings
shout = greeting.upper()
print("upper‑case:", shout)

# f‑string embeds values inside {}
age = 21
print(f"{name} is {age} years old.")

# Try changing words, slices, and formats to see different results.
```

## Chapter 6 – If and Match

This chapter introduces **decision making** in Python. You learn how to run different blocks of code
depending on conditions, and how Python decides what is considered “true”. You also see the newer
`match` statement for pattern matching.

**Key ideas from Chapter 6:**  
- Writing comments with `#` and breaking long lines with `\`  
- Using `if`, `elif`, and `else` to choose between paths  
- Truthiness: which values count as true or false in conditions  
- Checking membership with `in`  
- Using `match` to select actions based on structured patterns


```python
# Chapter 6 – making decisions with if and match

# A comment starts with # and is ignored by Python.

# Basic if/elif/else
temperature = 18

if temperature < 0:
    print("It is freezing!")
elif temperature < 20:
    print("It is a bit chilly.")
else:
    print("Nice and warm.")

# Truthiness: empty containers are False, non‑empty are True
items = []
if items:
    print("We have some items.")
else:
    print("The list is empty.")

# Membership with 'in'
name = "Ada"
if "d" in name.lower():
    print("The letter 'd' is in the name.")

# match statement (pattern matching)
status_code = 404

match status_code:
    case 200:
        print("OK")
    case 404:
        print("Not found")
    case 500:
        print("Server error")
    case _:
        print("Some other status")

# Try changing temperature, items, or status_code and see how the output changes.
```

## Chapter 7 – For and While

This chapter shows how to **repeat** code using loops. A `while` loop repeats while a condition is true,
and a `for` loop iterates over items in a sequence. You also see how to control loops with `break` and
`continue`, and how to use `range()` to generate number sequences.

**Key ideas from Chapter 7:**  
- `while` loops for “repeat while this condition holds”  
- `for` loops for stepping through items in order  
- Stopping early with `break` and skipping one step with `continue`  
- Optional `else` blocks on loops  
- Using `range()` to produce simple number sequences


```python
# Chapter 7 – loops with while and for

# while loop: countdown
countdown = 5
while countdown > 0:
    print("Counting down:", countdown)
    countdown -= 1
print("Blast off!")

print("-" * 30)

# for loop over a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print("I like", fruit)

print("-" * 30)

# Using break and continue
for number in range(1, 11):  # numbers 1 through 10
    if number == 5:
        print("Skipping 5")
        continue        # skip printing 5
    if number == 9:
        print("Stopping at 9")
        break           # stop the loop completely
    print("number:", number)

print("-" * 30)

# for with else: the else runs only if we didn't hit break
for n in range(3):
    print("n is", n)
else:
    print("Loop finished without break.")

# Try changing the ranges or conditions to practice.
```

## Chapter 8 – Tuples and Lists

This chapter introduces two important **sequence** types: tuples and lists. Both store ordered collections
of items. Tuples are fixed (immutable), while lists can be changed (mutable). You practice creating,
indexing, slicing, and modifying lists, and see list comprehensions for building lists quickly.

**Key ideas from Chapter 8:**  
- Creating tuples with commas and `()` and lists with `[]`  
- Accessing items by index and by slices  
- List methods such as `append()`, `insert()`, `remove()`, and `sort()`  
- Copying lists and understanding how assignment works  
- Using `zip()` to iterate over multiple sequences together  
- List comprehensions for compact list creation


```python
# Chapter 8 – tuples and lists

# A tuple is an ordered, immutable sequence
coordinates = (10.0, 20.0)
print("coordinates tuple:", coordinates)
print("x value:", coordinates[0])

# A list is an ordered, mutable sequence
animals = ["cat", "dog", "hamster"]
print("original animals:", animals)

# Indexing and slicing
print("first animal:", animals[0])
print("last two animals:", animals[1:])

# Modifying the list
animals.append("parrot")        # add to the end
animals.insert(1, "turtle")     # insert at position 1
print("after append and insert:", animals)

animals.remove("dog")           # remove by value
print("after removing 'dog':", animals)

animals.sort()                  # sort alphabetically
print("sorted animals:", animals)

# List comprehension – build a list from a pattern
numbers = [1, 2, 3, 4, 5]
squares = [n * n for n in numbers]   # square each number
print("numbers:", numbers)
print("squares:", squares)

# Using zip to iterate over two lists together
for animal, square in zip(animals, squares):
    print(animal, "paired with square", square)

# Try adding more animals or changing the comprehension expression.
```

## Chapter 9 – Sets and Dictionaries

This chapter covers **dictionaries** and **sets**, which store items by value rather than by position in a list.
Dictionaries map keys to values, and sets store unique, unordered items. These structures are very useful
for fast lookups and membership checks.

**Key ideas from Chapter 9:**  
- Creating dictionaries with `{}` or `dict()`  
- Adding, changing, and deleting entries by key  
- Looping over keys and values with `for` and `.items()`  
- Using sets for unique items and membership tests  
- Basic set operations like union, intersection, and difference  
- Dictionary and set comprehensions


```python
# Chapter 9 – dictionaries and sets

# A simple dictionary: language -> creator
creators = {
    "Python": "Guido van Rossum",
    "C": "Dennis Ritchie",
    "JavaScript": "Brendan Eich"
}

print("Python was created by:", creators["Python"])

# Adding or updating entries
creators["Go"] = "Robert Griesemer, Rob Pike, Ken Thompson"
creators["Python"] = "Guido van Rossum (and the community)"
print("Updated dictionary:", creators)

# Looping through keys and values
for language, creator in creators.items():
    print(f"{language} -> {creator}")

print("-" * 30)

# Using get() with a default value
print("Rust creator:", creators.get("Rust", "Unknown"))

# A set of unique values
colors = {"red", "green", "blue", "red"}  # duplicates are ignored
print("colors set:", colors)

colors.add("yellow")
print("after adding yellow:", colors)

# Set operations
warm_colors = {"red", "orange", "yellow"}
cool_colors = {"blue", "green", "purple"}

print("union:", warm_colors | cool_colors)
print("intersection:", warm_colors & colors)
print("difference (warm but not colors):", warm_colors - colors)

# Try adding more keys to the dictionary or more items to the sets.
```

## Chapter 10 – Functions

This chapter shows how to group code into reusable **functions**. Functions take input values (arguments),
perform some work, and optionally return a result. You also see default arguments, keyword arguments,
and a first look at handling errors with exceptions.

**Key ideas from Chapter 10:**  
- Defining functions with `def` and calling them with `()`  
- Parameters, positional arguments, and keyword arguments  
- Returning values with `return` (or `None` when there is no explicit return)  
- Default parameter values  
- Simple error handling with `try` and `except`  
- Writing docstrings to document what a function does


```python
# Chapter 10 – defining and using functions

# A simple function with a parameter
def greet(name):
    """Print a friendly greeting."""
    print(f"Hello, {name}!")

greet("Ada")
greet("Guido")

# A function with a default parameter value
def make_ticket(name, priority="normal"):
    """Return a short description of a help‑desk ticket."""
    return f"Ticket for {name} (priority: {priority})"

print(make_ticket("Alice"))
print(make_ticket("Bob", priority="high"))  # keyword argument

# A function that returns a value
def add(a, b):
    """Return the sum of a and b."""
    return a + b

result = add(3, 4)
print("3 + 4 =", result)

# Simple error handling with try/except
def safe_divide(a, b):
    """Divide a by b, handling division by zero."""
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero!")
        return None

print("10 / 2 =", safe_divide(10, 2))
print("5 / 0 =", safe_divide(5, 0))

# Try writing your own small function, such as one that
# computes the area of a rectangle from width and height.
```

## Chapter 11 – Objects

This chapter introduces **objects** and **classes**. Objects bundle data (attributes) and behavior (methods)
together. Classes are blueprints for creating objects. You also see simple inheritance, where one class
builds on another.

**Key ideas from Chapter 11:**  
- Defining a class with `class` and creating instances  
- Using `__init__` to set up attributes when an object is created  
- Writing methods that use `self` to access data on the object  
- Simple inheritance: a child class can extend or customize a parent class  
- Using attributes and methods from instances


```python
# Chapter 11 – classes and objects

# A simple class definition
class Dog:
    """A very small Dog class with a name and age."""

    def __init__(self, name, age):
        self.name = name   # instance attribute
        self.age = age

    def bark(self):
        print(f"{self.name} says: Woof!")

    def describe(self):
        print(f"{self.name} is {self.age} years old.")


# Create some Dog objects (instances)
fido = Dog("Fido", 3)
luna = Dog("Luna", 5)

fido.bark()
fido.describe()

luna.bark()
luna.describe()

print("-" * 30)

# Inheritance: GuideDog extends Dog
class GuideDog(Dog):
    def __init__(self, name, age, owner):
        super().__init__(name, age)  # call parent __init__
        self.owner = owner

    def describe(self):
        # Override the parent's describe method
        print(f"{self.name} is a guide dog for {self.owner}.")


buddy = GuideDog("Buddy", 4, "Sam")
buddy.bark()       # inherited from Dog
buddy.describe()   # overridden version

# Try adding your own method or attribute to Dog or GuideDog.
```

## Chapter 12 – Modules and Packages

This chapter shows how to organize code into **modules** and **packages** so that programs can grow
without becoming one huge file. You learn how to import existing modules from the standard library,
how to give them shorter aliases, and how modules make it easy to reuse code.

**Key ideas from Chapter 12:**  
- A module is a Python file that can be imported from other code  
- Importing with `import module`, `import module as name`, and `from module import thing`  
- The module search path and the idea of packages as directories of modules  
- Using helpful standard library modules  
- The idea of third‑party packages that you can install and then import


```python
# Chapter 12 – modules and imports

# Import a whole module
import math
print("pi from math module:", math.pi)
print("sin(pi/2):", math.sin(math.pi / 2))

# Import a single name from a module
from random import choice

options = ["rock", "paper", "scissors"]
print("Random choice:", choice(options))

# Import with an alias
import statistics as stats

numbers = [10, 20, 20, 30, 40]
print("mean:", stats.mean(numbers))
print("median:", stats.median(numbers))

# Example of how your own module might look (this would normally be in its own file):
#
#   # in file helpers.py
#   def shout(text):
#       return text.upper() + "!"
#
# and then in another file you could write:
#
#   from helpers import shout
#   print(shout("python"))
#
# Try exploring other standard library modules such as datetime or pathlib.
```
